```python
def countdown(n):
    print("start")
    while n>0:
        yield n
        n -= 1
    print("Done")

gen = countdown(3)

print(next(gen))
print(next(gen))

print(next(gen))

print(next(gen))


```

```python
def squares(n):
    for i in range(1, n+1):
        yield i**2

for s in squares(4):
    print(s)

```
