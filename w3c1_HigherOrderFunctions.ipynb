```python
def calculator(operation, num1 , num2):
    return operation(num1, num2)

def addition(n1,n2):
    return n1+n2

def subtraction(n1,n2):
    return n1-n2

def multiplication(n1,n2):
    return n1*n2

def division(n1,n2):
    return n1/n2

print(calculator(addition, 2,4))
print(calculator(subtraction, 2,4))
print(calculator(multiplication, 2, 4))
print(calculator(division, 2, 4))
```

```python
def multiplyer(n):
    def create_multiplyer(num):
        return n * num
    return create_multiplyer

two_multiplyer = multiplyer(2)
three_multiplyer = multiplyer(3)


print(multiplyer(2)(20))
print(two_multiplyer(10))
print(three_multiplyer(20))
```

```python
def greet(str):
    def greeting(st):
        return str + " "+ st
    return greeting

hi_greeting = greet("Hello")
hello_greeting = greet("Hi")

print(hi_greeting("Sir"))
print(hello_greeting("maam"))
```
