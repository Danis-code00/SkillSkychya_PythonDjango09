```python
class Crow:
    def Speak(self):
        print("Caw!")
class Parrot:
    def Speak(self):
        print("Squawk!")
class Duck:
    def Speak(self):
        print("Quack!")

for birds in [Crow(), Parrot(), Duck()]:
    birds.Speak()
```

```python
class Shape:
    def area(self):
        return 0
class Rectangle(Shape):
    def __init__(self, w, h):
        self.w, self.h = w, h
    def area(self):
        return self.w * self.h
class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return round(3.14 * self.r ** 2, 1)

class Square(Shape):
    def __init__(self,s):
        self.s = s
    def area(self):
        return self.s ** 2

for s in [Rectangle(3, 4), Circle(3), Square(5)]:
    print(s.area())

```

```python
class Money:
    def __init__(self, amount):
        self.amount = amount

    def __add__(self, other):
        return Money(self.amount + other.amount)

    def __sub__(self, other):
        return Money(self.amount - other.amount)

    def __mul__(self, other):
        return Money(self.amount * other.amount)

    def __truediv__(self, other):
        return Money(self.amount / other.amount)


add = Money(500) + Money(250)
sub = Money(500) - Money(250)
mul = Money(500) * Money(2)
div = Money(500) / Money(2)

print(add.amount)  # Output: 750
print(sub.amount)  # Output: 250
print(mul.amount)  # Output: 1000
print(div.amount)  # Output: 250
```
