```python
results = {
    'John': [85, 90, 78],
    'Alice': [92, 88, 95],
    'Bob': [78, 82, 88]
}

```

```python
#use membership operator to check iff alice is in dictonart
print(results['Alice'])
```

```python
avrage_marks = {}
for r in results:
   avrage = sum(results[r]) / len(results[r])
   avrage_marks[r] = avrage

print(avrage_marks)
```

```python
for key in results:
    results[key].sort
print(results)
```

```python
arranged_results = {}

for key in results:
    arranged_results[key] = sorted(results[key])

print(arranged_results)
```
