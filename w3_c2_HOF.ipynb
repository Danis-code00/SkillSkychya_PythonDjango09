```python
def apply_twice(func, x):
    def lamb(n):
        return func(func(n));
    return lamb(x)

apply_twice(lambda n:n+3 , 10)
```

```python
def make_greeting(word):
    def greet(name):
        return f"{word}, {name}!"
    return greet

def show_call(func):
    return lambda word: lambda name: f'{func.__name__}("{word}")("{name}")'

output = show_call(make_greeting)("Namaste")("Ram")

print(output)
```
