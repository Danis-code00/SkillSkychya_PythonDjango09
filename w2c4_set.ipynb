```python
#with list LATO method
dedupe =([3,1,3,2,1])
print((dedupe))
fix = []
for item in dedupe:
    if item not in fix:
        fix.append(item)
print(fix)
```

```python
# with set (Batho method)
dedupe =([3,1,3,2,1])
print((dedupe))
fix = []
efficient = set()
for item in dedupe:
    if item not in efficient:
        efficient.add(item)
        fix.append(item)
print(fix)
```
