```python
def sq(n):
    return n*n

results = map(sq, [1,2,3,4,5])

print(results)
print(list(results))
```

```python
def even(n):
    return n%2 == 0

results = filter(even,[1,2,3,4,5])

print(list(results))
```

```python
from functools import reduce
def find_sum(x,y):
    return x+y

results = reduce(find_sum,[1,2,3,4,5])
print(results)
```

```python
def find_max(x,y):
    if x>y:
        return x
    return y

results = reduce (find_max, [1,2,3,4,5])

print(results)
```

```python
results = map (lambda x: x**2, [1,2,3,4,5])

print(list(results))
```

```python
cube = lambda x: x**3;
print(cube(3))
```

```python
temps = [0,25,100]
f = lambda x : x*9/5 + 32
print(list(map(f, temps)))
```

```python
names = ["Ram", "Sita", "Hari", "Om"]

print(list(filter(lambda x: len(x) >3, names)))
```
