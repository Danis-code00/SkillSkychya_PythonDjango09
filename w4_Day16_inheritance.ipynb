```python
class Vehicle:
    def __init__(self, brand, wheels):
        self.brand = brand
        self.wheels = wheels
        self.has_carrier = False

    def info(self):
        if self.has_carrier:
            return f"{self.brand} with {self.wheels} (carrier: Yes)"
        return f"{self.brand} with {self.wheels} wheels (carrier: No)"


class Car(Vehicle):
    def __init__(self, brand):
        super().__init__(brand, wheels=4)


class Bike(Vehicle):
    def __init__(self, brand, has_carrier=False):
        super().__init__(brand, wheels=2)
        self.has_carrier = has_carrier

class ElectricCar(Car):
    def __init__(self,brand, battery_kwh):
        super().__init__(brand)
        self.battery_kwh = battery_kwh


c = Car("Toyota")
b = Bike("Honda", has_carrier=True)
e = ElectricCar("Tesla", battery_kwh=100)

print(c.info())
print(b.info())
print(e.info())
```
